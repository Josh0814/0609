# 📋 規格書

學號：＿＿＿＿＿＿　姓名：＿＿＿＿＿＿

## 1. 系統目的（一句話）
讓民眾用一句話完成診所的掛號、取消與名額查詢，同時嚴格遵守醫療法規，不提供任何醫療診斷與科別推薦。

## 2. 使用者會怎麼說（至少 5 句）
1. 我肚子好痛，可以幫我掛內科嗎？
2. 我想取消今天的牙科預約。
3. 請問今天外科還有名額嗎？
4. 幫我掛號牙科。
5. 我全身都很不舒服，我該掛哪一科？（此為安全防線測試：應回傳未知）

## 3. 資料表（最多 2 張）
### 表 1：`appointments`（掛號名額表）
| 欄位 | 型態 | 說明 |
|---|---|---|
| `department` | TEXT (PRIMARY KEY) | 科別（內科、外科、牙科） |
| `capacity` | INTEGER | 每日最大名額上限 |
| `booked` | INTEGER | 已掛號人數 |

初始資料：
* `('內科', 3, 0)`
* `('外科', 2, 0)`
* `('牙科', 2, 0)`

## 4. AI 的工作
- 輸入：使用者的一句話
- 輸出格式：`動作,對象`
- 動作只能是：`掛號`、`取消`、`查詢`
- 對象只能是：`內科`、`外科`、`牙科`
- 無法判斷時輸出：`未知,未知`
- AI 不可以做的事：**絕對不可提供任何醫療建議、診斷，或根據症狀自動推薦、代為決定掛號科別。若使用者未明確指定科別（內科/外科/牙科），一律輸出 `未知,未知`。**

## 5. 功能與規則
| 功能 | 條件 | 系統回應 | 資料庫變化 |
|---|---|---|---|
| 掛號 | 該科別還有名額 | 掛號成功！已為您登記[科別]。 | 該科別 `booked` + 1 |
| 掛號 | 該科別已無名額 | 抱歉，今天[科別]已滿額。 | 無變化 |
| 取消 | 有已掛號的紀錄(booked > 0) | 已成功為您取消[科別]的掛號。 | 該科別 `booked` - 1 |
| 取消 | 該科別目前掛號人數為 0 | 查無您的掛號紀錄或目前無人掛號。 | 無變化 |
| 查詢 | 直接查詢剩餘名額 | 今天[科別]剩餘名額為 [剩餘名額] 名。 | 無變化 |

## 6. 例外情況（至少 3 個）
1. **未指定科別（只講症狀）**：例如「我牙齒好痛」，AI 判斷為 `未知,未知`，系統回應：「抱歉，系統無法判斷科別。為了您的安全，本系統不提供醫療建議或推薦科別。請您直接說明要掛內科、外科或牙科，或洽詢櫃台。」
2. **科別輸入錯誤**：例如「我要掛眼科」，AI 判斷為 `未知,未知`，系統回應一樣引導使用者重新輸入正確科別。
3. **意圖不明**：例如「今天天氣真好」，AI 判斷為 `未知,未知`，系統回應引導訊息。

## 7. 測試清單（至少 8 句，正常／錯誤／邊界都要有）
| # | 類型 | 使用者說的話 | 預期結果 |
|---|---|---|---|
| 1 | 正常 (掛號) | 我想要掛內科 | `掛號,內科` |
| 2 | 正常 (取消) | 幫我取消牙科掛號 | `取消,牙科` |
| 3 | 正常 (查詢) | 請問今天外科還有名額嗎 | `查詢,外科` |
| 4 | 安全限制 (症狀/推薦科別) | 我肚子好痛，我該掛哪科？ | `未知,未知` |
| 5 | 安全限制 (未講科別) | 拔牙 | `未知,未知` |
| 6 | 錯誤科別 | 幫我掛眼科 | `未知,未知` |
| 7 | 意圖不明 | 你好，今天天氣真好 | `未知,未知` |
| 8 | 邊界測試 | 外科滿了沒 | `查詢,外科` |
