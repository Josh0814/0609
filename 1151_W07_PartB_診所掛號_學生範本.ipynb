````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型
# ============================================================
!pip install -q groq gradio

import sqlite3
import pandas as pd
from IPython.display import display
from google.colab import userdata
from groq import Groq
import gradio as gr

# 🔴 金鑰只能從左側「🔑 密鑰（Secrets）」讀取，絕對不可以寫成 api_key="gsk_..."
#    寫死的金鑰一旦存到 GitHub，就等於公開給全世界使用。
#    也不要 print 出金鑰，因為輸出結果也會一起存到 GitHub。
try:
    client = Groq(api_key=userdata.get('GROQ_API_KEY'))
except Exception as e:
    print("❌ 讀不到金鑰。請檢查：")
    print("   1. 左側 🔑 是否新增了名稱為 GROQ_API_KEY 的密鑰（大小寫要完全一樣）")
    print("   2. 該密鑰的「筆記本存取權」開關是否已打開")
    raise

# 模型名稱集中放在這裡；日後模型更換或下架，只要改這一行
MODEL = "openai/gpt-oss-120b"
print("✅ Groq 金鑰讀取成功，使用模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# ============================================================
import sqlite3

def init_db():
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 每次執行都先 DROP TABLE 以便重置資料
        cursor.execute("DROP TABLE IF EXISTS appointments")
        # 建立 appointments 資料表
        cursor.execute("""
            CREATE TABLE appointments (
                department TEXT PRIMARY KEY,
                capacity INTEGER,
                booked INTEGER
            )
        """)
        # 插入初始資料
        initial_data = [
            ('內科', 3, 0),
            ('外科', 2, 0),
            ('牙科', 2, 0)
        ]
        cursor.executemany("INSERT INTO appointments VALUES (?, ?, ?)", initial_data)
        conn.commit()
        print("✅ 資料庫與初始資料建立成功！")
    finally:
        conn.close()

# 執行初始化
init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# ============================================================
import sqlite3
import pandas as pd
from IPython.display import display

def show_tables():
    conn = sqlite3.connect('clinic.db')
    try:
        # 讀取資料表內容並以 DataFrame 漂亮顯示
        df = pd.read_sql_query("SELECT * FROM appointments", conn)
        display(df)
    finally:
        conn.close()

# 呼叫一次顯示初始狀態
print("--- 目前診所掛號名額狀態 ---")
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# ============================================================
import json

def analyze_user_input(user_text):
    # 限制動作與對象的允許清單
    valid_actions = ['掛號', '取消', '查詢']
    valid_departments = ['內科', '外科', '牙科']

    prompt = f"""你是一個診所掛號系統的分類助理。
你的工作是分析使用者的話，並嚴格按照格式回傳。

【限制與安全規定】
1. 動作只能是：掛號、取消、查詢
2. 對象只能是：內科、外科、牙科
3. 絕對不可提供任何醫療建議、診斷，或根據症狀自動推薦、代為決定掛號科別。
4. 若使用者未明確指定科別（例如：只說肚子痛、拔牙、或不相干話題），一律輸出「未知,未知」。

【輸出格式】
動作,對象

【範例】
使用者：我要掛內科 -> 掛號,內科
使用者：幫我取消牙科 -> 取消,牙科
使用者：外科還有名額嗎 -> 查詢,外科
使用者：我頭痛要看哪一科？ -> 未知,未知

使用者輸入：{user_text}
"""

    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=0
        )
        raw_reply = response.choices[0].message.content.strip()

        # 解析 AI 的原始回覆
        parts = raw_reply.split(',')
        if len(parts) == 2:
            action = parts[0].strip()
            dept = parts[1].strip()
        else:
            action, dept = "未知", "未知"

        # 嚴格驗證是否在允許清單內
        if action not in valid_actions:
            action = "未知"
        if dept not in valid_departments:
            dept = "未知"

        return action, dept, raw_reply
    except Exception as e:
        # 發生異常時不崩潰，回傳未知
        return "未知", "未知", f"Error: {str(e)}"
````

````python
# ============================================================
# 第 5 格：測試清單
# ============================================================

# 測試清單格式：(使用者說的話, 預期動作, 預期科別)
test_cases = [
    ("我想要掛內科", "掛號", "內科"),
    ("幫我取消牙科掛號", "取消", "牙科"),
    ("請問今天外科還有名額嗎", "查詢", "外科"),
    ("我肚子好痛，我該掛哪科？", "未知", "未知"),
    ("拔牙", "未知", "未知"),
    ("幫我掛眼科", "未知", "未知"),
    ("你好，今天天氣真好", "未知", "未知"),
    ("外科滿了沒", "查詢", "外科")
]

def run_tests():
    passed_count = 0
    print(f"開始測試... 總共 {len(test_cases)} 筆測試案例\n")
    print(f"{'#':<3} | {'輸入文字':<15} | {'預期結果':<10} | {'AI原始回覆':<12} | {'系統解讀':<10} | {'結果':<5}")
    print("-" * 80)

    for idx, (text, exp_act, exp_dept) in enumerate(test_cases, 1):
        act, dept, raw = analyze_user_input(text)
        expected = f"{exp_act},{exp_dept}"
        parsed = f"{act},{dept}"
        is_passed = (act == exp_act and dept == exp_dept)

        if is_passed:
            passed_count += 1
            status = "✅"
        else:
            status = "❌"

        print(f"{idx:<3} | {text:<15} | {expected:<10} | {raw:<12} | {parsed:<10} | {status}")

    print("-" * 80)
    print(f"測試結束：通過 {passed_count} / {len(test_cases)} 筆！")

# 執行測試
run_tests()
````

````python
# ============================================================
# 第 6 格：功能函式（掛號、取消、查詢）
# ============================================================
import sqlite3

def handle_registration(dept):
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 先查詢該科別目前的名額狀況
        cursor.execute("SELECT capacity, booked FROM appointments WHERE department = ?", (dept,))
        row = cursor.fetchone()
        if row is None:
            return f"抱歉，找不到「{dept}」這個科別。"

        capacity, booked = row
        if booked < capacity:
            # 還有名額，進行掛號 (booked + 1)
            cursor.execute("UPDATE appointments SET booked = booked + 1 WHERE department = ?", (dept,))
            conn.commit()
            return f"掛號成功！已為您登記[{dept}]。"
        else:
            return f"抱歉，今天[{dept}]已滿額。"
    finally:
        conn.close()

def handle_cancellation(dept):
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        cursor.execute("SELECT booked FROM appointments WHERE department = ?", (dept,))
        row = cursor.fetchone()
        if row is None:
            return f"抱歉，找不到「{dept}」這個科別。"

        booked = row[0]
        if booked > 0:
            # 有掛號紀錄，進行取消 (booked - 1)
            cursor.execute("UPDATE appointments SET booked = booked - 1 WHERE department = ?", (dept,))
            conn.commit()
            return f"已成功為您取消[{dept}]的掛號。"
        else:
            return f"查無您的掛號紀錄或目前無人掛號。"
    finally:
        conn.close()

def handle_query(dept):
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        cursor.execute("SELECT capacity, booked FROM appointments WHERE department = ?", (dept,))
        row = cursor.fetchone()
        if row is None:
            return f"抱歉，找不到「{dept}」這個科別。"

        capacity, booked = row
        remaining = capacity - booked
        return f"今天[{dept}]剩餘名額為 {remaining} 名。"
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與完整流程測試
# ============================================================

def process_clinic_system(user_text):
    # 1. 先由 AI 解讀意圖
    action, dept, raw = analyze_user_input(user_text)

    # 2. 如果無法解讀，執行安全例外防線與友善提示
    if action == "未知" or dept == "未知":
        return "抱歉，系統無法判斷科別。為了您的安全，本系統不提供任何醫療建議或科別推薦。請您直接說明要掛內科、外科或牙科，或洽詢櫃台。"

    # 3. 根據解讀結果執行對應功能
    if action == "掛號":
        return handle_registration(dept)
    elif action == "取消":
        return handle_cancellation(dept)
    elif action == "查詢":
        return handle_query(dept)
    else:
        return "系統發生未知錯誤，請稍後再試。"

# --- 模擬真實操作流程測試 ---
print("=== 模擬操作流程測試 ===\n")
init_db() # 重新初始化

# 1. 查詢名額 (剩餘 3 名)
print("使用者：『請問內科滿了嗎？』")
print("系統：", process_clinic_system("請問內科滿了嗎？"))
print("")

# 2. 連續掛號 3 次 (內科上限 3)
for i in range(1, 4):
    print(f"使用者：『我要掛號內科 ({i})』")
    print("系統：", process_clinic_system("我要掛號內科"))
print("")

# 3. 第 4 次掛號應失敗
print("使用者：『再幫我掛一次內科』")
print("系統：", process_clinic_system("再幫我掛一次內科"))
print("")

# 4. 取消掛號
print("使用者：『幫我取消內科預約』")
print("系統：", process_clinic_system("幫我取消內科預約"))
print("")

# 5. 再次查詢剩餘名額 (應回歸 1)
print("使用者：『幫我查內科還有多少名額』")
print("系統：", process_clinic_system("幫我查內科還有多少名額"))
print("")

# 6. 安全例外防線測試
print("使用者：『我牙齒好痛痛到不行』")
print("系統：", process_clinic_system("我牙齒好痛痛到不行"))
print("\n=== 最終資料庫狀態 ===")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# ============================================================
import gradio as gr

def gradio_interface(user_message, chat_history):
    # 呼叫主流程處理函數
    response = process_clinic_system(user_message)

    # 更新對話歷史
    chat_history.append((user_message, response))
    return "", chat_history

with gr.Blocks(title="安心診所掛號系統") as demo:
    gr.Markdown("## 🏥 安心診所掛號系統 (Clinic Registration System)")
    gr.Markdown("歡迎使用語音/文字智慧掛號，您可以對我說：\n"
                "* **掛號**：『我想掛內科』、『幫我掛號牙科』\n"
                "* **取消**：『取消明天的外科預約』\n"
                "* **查詢**：『請問牙科還有名額嗎』\n\n"
                "*⚠️ 本系統遵照法規不提供任何醫療建議與診斷，亦不代為推薦或決定科別。*")

    chatbot = gr.Chatbot(label="對話紀錄")
    msg = gr.Textbox(label="請輸入您的需求（例如：我要掛號內科）", placeholder="在此輸入文字...")
    clear = gr.Button("清除對話")

    # 綁定事件
    msg.submit(gradio_interface, [msg, chatbot], [msg, chatbot])
    clear.click(lambda: None, None, chatbot, queue=False)

# 啟動 Gradio，設定 share=True 方便外部測試
demo.launch(share=True)
````
